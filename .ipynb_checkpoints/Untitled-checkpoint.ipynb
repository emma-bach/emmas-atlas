```python
import matplotlib.pyplot as plt
import numpy as np

xarr = yarr = [i/100 for i in range(-100,100)]
X,Y = np.meshgrid(xarr, yarr)

f = lambda xgrid,ygrid : np.array([[(x * y)/(x**2 + y**2)**2 if (x,y) != (0,0) else 0 for x in xgrid] for y in ygrid])
Z = f(xarr,yarr)

fig = plt.figure()
ax = fig.add_subplot(projection = '3d')
Z[Z > 25] = np.nan
Z[Z < -25] = np.nan
ax.plot_surface(X,Y,Z)
import pickle
pickle.dump(fig, open('FigureObject.fig.pickle', 'wb'))
```
